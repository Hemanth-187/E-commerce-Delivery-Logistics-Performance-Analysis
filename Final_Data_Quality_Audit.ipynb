```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "DataCoSupplyChainDataset.csv",
    encoding="latin1"
)

print("Dataset loaded successfully")
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
```

```python
missing_audit = pd.DataFrame({
    "Missing Values Count": df.isnull().sum(),
    "Missing Values Percentage":
        (df.isnull().mean() * 100).round(2),
    "Non-Missing Values Count": df.notnull().sum()
})

missing_audit = missing_audit[
    missing_audit["Missing Values Count"] > 0
].sort_values("Missing Values Count", ascending=False)

display(missing_audit)
```

```python
print("Exact duplicate rows:", df.duplicated().sum())

print(
    "Repeated Order ID entries:",
    df["Order Id"].duplicated().sum()
)

order_counts = df["Order Id"].value_counts()
repeated_ids = order_counts[order_counts > 1]

print("Unique Order IDs repeated:", len(repeated_ids))
```

```python
# Check invalid or inconsistent values
print("Negative Sales:", (df["Sales"] < 0).sum())
print("Zero or negative Order Quantity:",
      (df["Order Item Quantity"] <= 0).sum())

# Check date validity
order_dates = pd.to_datetime(
    df["order date (DateOrders)"],
    errors="coerce"
)

shipping_dates = pd.to_datetime(
    df["shipping date (DateOrders)"],
    errors="coerce"
)

print("Invalid order dates:", order_dates.isna().sum())
print("Invalid shipping dates:", shipping_dates.isna().sum())
print("Shipping before order:",
      (shipping_dates < order_dates).sum())
```

```python
audit_summary = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Total Missing Values",
        "Columns with Missing Values",
        "Exact Duplicate Rows",
        "Repeated Order ID Entries",
        "Unique Order IDs Repeated"
    ],
    "Result": [
        len(df),
        len(df.columns),
        int(df.isnull().sum().sum()),
        int((df.isnull().sum() > 0).sum()),
        int(df.duplicated().sum()),
        int(df["Order Id"].duplicated().sum()),
        int((df["Order Id"].value_counts() > 1).sum())
    ]
})

display(audit_summary)
```
