```python
import pandas as pd

df = pd.read_csv("DataCoSupplyChainDataset.csv", encoding="latin1")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
missing_counts = df.isnull().sum()
missing_counts = missing_counts[missing_counts > 0]

print(missing_counts)
```

```python
missing_counts = df.isnull().sum()
missing_counts = missing_counts[missing_counts > 0]

print(missing_counts)
```

```python
import pandas as pd

missing_counts = df.isnull().sum()
missing_counts = missing_counts[missing_counts > 0]

missing_log = pd.DataFrame({
    "Column_Name": missing_counts.index,
    "Missing_Count": missing_counts.values
})

rules = {
    "Customer Lname": "Flag for review; do not guess the name",
    "Customer Zipcode": "Flag for review; do not guess the zipcode",
    "Order Zipcode": "Record as missing; review before geographic analysis",
    "Product Description": "Keep missing; do not invent descriptions"
}

missing_log["Handling_Rule"] = missing_log["Column_Name"].map(rules)
missing_log["Action_Taken"] = "Missing values identified; no values changed"

missing_log.to_csv("missing_data_handling_log.csv", index=False)

display(missing_log)
print("Missing-data handling log created successfully!")
```

```python
special_records = df[df.isnull().any(axis=1)].copy()

special_records.to_csv("special_treatment_records.csv", index=False)

print("Records requiring review:", len(special_records))
print("Special-treatment records file created!")
```
