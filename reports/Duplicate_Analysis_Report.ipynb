```python
df = pd.read_csv(
    "DataCoSupplyChainDataset.csv",
    encoding="latin1"
)
print("Dataset shape:", df.shape)
df.head()
```

# Exact duplicate rows

```python
print("Exact duplicate rows:", df.duplicated().sum())
```

# Repeated Order ID entries

```python
print(
    "Repeated Order ID entries:",
    df["Order Id"].duplicated().sum()
)
```

# Unique Order IDs that occur more than once

```python
order_counts = df["Order Id"].value_counts()
repeated_ids = order_counts[order_counts > 1]

print("Unique Order IDs repeated:", len(repeated_ids))
print(repeated_ids.head(10))
```
