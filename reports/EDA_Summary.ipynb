# Exploratory Data Analysis Summary

## Scope

This summary reflects the cleaned/interim DataCo supply-chain dataset at
`data/interim/DataCoSupplyChain_M5_cleaned.csv` and the analysis in
[`notebooks/Initial_EDA.ipynb`](./notebooks/Initial_EDA.ipynb). The dataset
contains 180,519 order-item rows, 53 columns, and 65,752 distinct orders
(approximately 2.75 rows per order). The recorded order-date range is January
2015 through January 2018. Percentages and rates below are row-level unless
otherwise stated.

## Key findings

1. **Late delivery is the most common delivery status.** There are 98,977 rows
   labeled `Late delivery` (54.8%), 41,592 `Advance shipping` (23.0%), 32,196
   `Shipping on time` (17.8%), and 7,754 `Shipping canceled` (4.3%).

2. **Duration distributions differ, and cancellations affect the comparison.**
   Among the 172,765 rows not labeled `Shipping canceled`, actual shipping time
   averages 3.50 days versus 2.93 scheduled days. In this subset, actual time
   exceeds the scheduled duration on 98,977 rows, matches it on 32,196, and is
   shorter on 41,592. These counts align with the late, on-time, and advance
   delivery labels, respectively. A second pair of histograms in the notebook
   includes all 180,519 rows, including canceled shipments; interpret it
   separately.

3. **Late-delivery rates vary by shipping mode.** Row-level rates are 95.3% for
   First Class (26,513/27,814), 76.6% for Second Class (26,987/35,216), 45.7%
   for Same Day (4,454/9,737), and 38.1% for Standard Class (41,023/107,752).
   Standard Class has the lowest rate but the greatest number of late rows
   because its group is much larger. Rates may also reflect different
   scheduled windows, routes, or customer/order mixes; this is a descriptive
   association, not a like-for-like causal comparison.

4. **Regional late rates are more tightly clustered than shipping-mode rates
   in the plotted comparison.** Central Africa has the highest rate in the top
   ten regions shown (58.0%; 972/1,677). Those displayed regions are mostly
   between 55.3% and 56.3%. Region counts differ substantially; for example,
   Central Asia has 553 rows and Western Europe has 27,109.

5. **Monthly order counts rise near the end, alongside a marked change in
   item coverage.** Monthly distinct orders range from 1,585 (February 2015)
   to 2,124 (December 2017). Average item rows per order are about 3.00 through
   September 2017, then 1.07 in October and 1.00 from November 2017 through
   January 2018. This could reflect a coverage or composition change; the
   available data alone does not establish its cause. Treat late-period row
   volumes and item-level financial comparisons with particular caution.

6. **Order-level and item-level delivery rates are similar, but use different
   denominators.** There are 36,048 late orders out of 65,752 (54.8%), versus
   98,977 late item rows out of 180,519 (54.8%). Cancellations are 2,855 orders
   (4.3%) versus 7,754 item rows (4.3%). Delivery status, order status, shipping
   mode, and actual/scheduled durations are consistent across all items within
   each order. Report the grain and denominator with every KPI.

7. **Order status and delivery status describe different outcomes.** All
   2,855 distinct orders labeled `Shipping canceled` have order status
   `CANCELED` (1,367) or `SUSPECTED_FRAUD` (1,488). Other order statuses include
   late, on-time, and advance-shipping outcomes, so do not treat an order
   status such as `COMPLETE` as a substitute for delivery performance.

8. **Shipping-mode comparisons are confounded with the scheduled window.** Each
   shipping mode has exactly one scheduled-duration value in this data:
   Same Day=0, First Class=1, Second Class=2, Standard Class=4 days. There is
   no within-schedule overlap for a like-for-like mode comparison; observed
   rate differences cannot isolate a shipping-mode effect.

9. **Recorded date gaps mostly agree with actual duration, with exceptions.**
   The calendar-day gap from order timestamp to shipping timestamp differs
   from `Days for shipping (real)` on 4,657 rows (2.6%): 4,454 late-delivery
   rows and 203 canceled rows. Investigate timestamp/calendar conventions and
   definitions before changing values.

10. **Monthly late rates vary without a clear sustained trend.** Distinct-order
    late rates range from 52.45% (July 2016) to 57.28% (June 2016) across 37
    months; monthly cancellation rates range from 3.35% to 5.46%. The short
    period and month-to-month variation do not establish a trend.

11. **Order benefit is positive on average but includes negative values.**
   `Benefit per order` averages 21.97, with a median of 31.52; 33,784 rows
   (18.7%) are negative. `Benefit per order` and `Order Profit Per Order` are
   identical on every row, so they are not independent measures. Market-level
   medians range from about 29.18 to 32.78, and negative shares are about
   18.6%-18.9%. Verify metric definitions and currency before interpreting
   these values as profitability.

12. **The late-delivery risk flag duplicates the delivery outcome.**
   `Late_delivery_risk` matches whether `Delivery Status` is `Late delivery`
   on all 180,519 rows. This is a consistency observation, not an independent
   predictor of the outcome.

## Data limitations

- **Order-item grain:** Rows represent order items, not unique orders.
  Order-level and row-level rates happen to be similar here because delivery
  outcomes are consistent within each order; denominators still differ, and
  item rows within an order are not independent.
- **Missing fields:** `Product Description` is missing for all 180,519 rows;
  `Order Zipcode` is missing for 155,679 rows (86.2%). `Customer Lname` has 8
  missing values and `Customer Zipcode` has 3. This limits product-description
  and detailed location analysis.
- **Delivery-status interpretation:** The duration/status alignment above
  excludes rows labeled `Shipping canceled`. Confirm operational definitions
  and keep cancellations separate. A non-canceled delivery status does not
  necessarily mean the order is complete; the dataset also has a separate
  `Order Status` field. Calendar-day timestamp gaps differ from recorded
  actual duration on 4,657 rows, which needs definition-level investigation.
- **Target leakage:** Because `Late_delivery_risk` exactly matches the
  `Late delivery` label, it must not be used to predict that outcome. A
  prospective model should use only information available before delivery.
- **Unequal group sizes and confounding:** Region and shipping-mode groups have
  different sizes. Each shipping mode maps to a single scheduled duration in
  this sample, so mode and promised window cannot be separated by
  stratification. Smaller-group rates are less stable and should be
  interpreted with counts and uncertainty, not rankings alone.
- **Limited temporal coverage:** The dataset spans January 2015 through
  January 2018. The sharp change in item rows per order from October 2017
  onward may indicate a coverage/composition change; its cause is unknown,
  and trends may not generalize beyond this period.
- **Financial interpretation:** Currency and calculation definitions for
  `Benefit per order` have not been established here; it duplicates
  `Order Profit Per Order`, and negative/extreme values can influence
  summary statistics.
- **Privacy:** The source includes customer-identifying and credential-like
  fields. Do not expose these fields in reports or visualizations.
- **Descriptive analysis:** The observed differences are associations and do
  not establish causes or guarantee generalizability.

## Recommended next steps

- Report both order-item and distinct-order delivery KPIs, and document each
  aggregation rule and denominator.
- Validate delivery and order-status definitions; keep canceled rows separate
  and investigate the 4,657 date-gap/duration mismatches.
- Investigate the late-2017 reduction in item rows per order before
  comparing item-level volume or financial metrics over time.
- Avoid interpreting shipping-mode differences as mode effects: each mode has
  only one scheduled-duration value in this sample. Seek data with overlapping
  mode/window combinations, or describe current results as unadjusted rates.
- Exclude `Late_delivery_risk` from predictive features for delivery outcome
  and use a time-aware validation approach.
- Confirm financial metric definitions, currency, and extreme values before
  making profitability recommendations.
- Document missingness and privacy controls, then select validated KPIs for
  the next analysis phase.
