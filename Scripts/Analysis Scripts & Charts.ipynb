```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Dataset path
file_path = Path(r"C:\Users\HP\Internship\DataCoSupplyChainDataset.csv")

# Load dataset
df = pd.read_csv(file_path, encoding="cp1252")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
#Delivery Status Cah
plt.figure(figsize=(9, 5))

df["Delivery Status"].value_counts().plot(kind="bar")

plt.title("Delivery Status Analysis")
plt.xlabel("Delivery Status")
plt.ylabel("Number of Orders")
plt.xticks(rotation=30)
plt.tight_layout()

plt.savefig(chart_path / "01_delivery_status.png")
plt.show()
plt.close()
```

```python
# 2. Shipping Mode Analysis
shipping = df.groupby("Shipping Mode")[
    ["Days for shipping (real)",
     "Days for shipment (scheduled)"]
].mean()

shipping.plot(kind="bar", figsize=(9, 5))

plt.title("Actual vs Scheduled Shipping Days")
plt.xlabel("Shipping Mode")
plt.ylabel("Average Days")
plt.xticks(rotation=30)
plt.tight_layout()

plt.savefig(chart_path / "02_shipping_mode.png")
plt.show()
plt.close()
```

```python
# 3. Monthly Orders Trend
df["order date (DateOrders)"] = pd.to_datetime(
    df["order date (DateOrders)"],
    errors="coerce"
)

df["Month"] = df["order date (DateOrders)"].dt.to_period("M")

monthly = df.groupby("Month")["Order Id"].nunique()

monthly.plot(kind="line", marker="o", figsize=(10, 5))

plt.title("Monthly Orders Trend")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(chart_path / "03_monthly_orders.png")
plt.show()

plt.close()
```

```python
# 4. Late Delivery Risk
risk = df["Late_delivery_risk"].value_counts().sort_index()

plt.figure(figsize=(7, 5))

risk.plot(kind="bar")

plt.title("Late Delivery Risk Analysis")
plt.xlabel("Late Delivery Risk (0 = No, 1 = Yes)")
plt.ylabel("Number of Orders")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig(chart_path / "04_delay_risk.png")
plt.show()
plt.close()

print("All four charts generated successfully!")
```
